# How do I Open these .ipynb files?

- These are Python Jupyter Notebooks, or iPython files
- I like them because I can multipurpose slides for information and run code in them at the same time!
- Researchers like them for a variety of reasons:
    - Interactive Exploration: You can run a code line by line (or cell by cell) which is perfect for testing out individual hypotheses, tweaking parameters, and inspecting results
    - It supports inline visualisation, like Matplotlib plots and Dataframe snapshots
    - It includes “magic” commands like %timeit, %run, and %matplotlib that simplify profiling, debugging, and visualization


# How do I run these files?

You have several options, I'll run through them all quickly but you can choose your favourite.

## Option 1: Open them with Google Colab

- The upside: This is the easiest way, and doesn't require you to install anything. It's great for if you have a google account and are working on a public computer.
- The downside: It's a google product. 
- Optional: Clone or Download the Repository - don't worry if you're on a public computer, you don't have to do this.
- Go to https://colab.research.google.com/ and sign into your google account.
- Click on File and New Notebook
- Now you can either upload the notebook you downloaded, or drop a link to the github.
- This copy is saved on your drive and you will be able to edit it at will.

## Option 2: Using Anaconda and Jupyter Notebooks.

- The upside: It's open source and not as evil as the other options.
- The downside: It's large, unwieldy, and kind of slow.
- Clone or Download the Repository
- Install [[Anaconda]](https://www.anaconda.com/download) on your machine
- Launch the navigator and start Jupyter Notebooks. 
- This will start your web browser on the localhost.
- Navigate to your downloaded repository and open the File in your browser.

## Option 3: Visual Studio Code

- The upside of Visual Studio Code is that you can use both iPython and scripts in the same place.
- The downside it that it's a microsoft product.
- Download https://code.visualstudio.com/download.
- It will run the file just like any python script.
