```python
file = open("../../data/story.txt", "r")  # 'r' for read
```

- Modes: "r" (read), "w" (write), "a" (append), "x" (create)

```python
with open("../../data/story.txt", "r") as file:
    content = file.read()
```

Methods to read data:
- read() – entire file
- readline() – one line
- readlines() – list of lines


Writing to a file

```python
with open("output.txt", "w") as file:
    file.write("Hello, world!")
```

Bonus tips

- Use try/except for error handling
- Check if file exists before opening
- Use .close() if not using with



```python
filename = "../../data/story.txt"

try:
    with open(filename, "r") as file:
        content = file.read()
        print(" File read successfully!")
except FileNotFoundError:
    print(f" File '{filename}' not found. Please check the filename or path.")
except Exception as e:
    print(f" An unexpected error occurred: {e}")
```
